```python
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

# 1. Upload & Import Data
files.upload()
df = pd.read_excel("Fahmi.xlsx")
df.columns = df.columns.str.strip()

# 2. Tabel Frekuensi
frq = df["Grade"].value_counts()
print("=== TABEL FREKUENSI ===\n", frq)

# 3. Visualisasi Data
plt.style.use('seaborn-v0_8-whitegrid')

# Line Chart
plt.figure(figsize=(6, 3))
plt.plot(frq.index, frq.values, marker='o', color='#4C78A8', lw=2)
plt.title('Distribusi Grade (Line)')
plt.show()

# Bar Chart
plt.figure(figsize=(6, 3))
plt.bar(frq.index, frq.values, color='#F58518')
plt.title('Distribusi Grade (Bar)')
plt.show()

# Pie Chart
plt.figure(figsize=(4, 4))
plt.pie(frq.values, labels=frq.index, autopct='%1.0f%%',
        colors=['#7B61A8', '#E45756', '#72B7B2', '#F2CF5B', '#54A24B'])
plt.title('Proporsi Grade (Pie)')
plt.show()

# 4. Statistika Deskriptif
col = [c for c in df.select_dtypes('number') if 'NIM' not in c.upper()][-1]
dt = pd.to_numeric(df[col], errors='coerce')

stats = dt.describe()
stats['Median'] = dt.median()
stats['Mode'] = dt.mode()[0]
stats['Variance'] = dt.var()
stats['Skewness'] = dt.skew()

print(f"\n=== STATISTIKA DESKRIPTIF ({col}) ===")
print(stats)
```
